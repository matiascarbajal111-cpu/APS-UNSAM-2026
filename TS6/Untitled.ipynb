## 1. Circuito Pasa-bajos
  $$Z_1(s) = sL + R$$

  $$Z_2(s) = \frac{1}{sC}$$

Aplicando el divisor de tensión para obtener $V_2(s) / V_1(s)$:

$$H(s) = \frac{V_2(s)}{V_1(s)} = \frac{Z_2(s)}{Z_1(s) + Z_2(s)} = \frac{\frac{1}{sC}}{(sL + R) + \frac{1}{sC}}$$

Multiplicando numerador y denominador por $sC$:

$$H(s) = \frac{1}{sC(sL + R) + 1} = \frac{1}{s^2 LC + sRC + 1}$$

Dividiendo numerador y denominador por $LC$ para expresarla en su forma canónica:

$$H(s) = \frac{\frac{1}{LC}}{s^2 + s\frac{R}{L} + \frac{1}{LC}}$$



## 1.2. Reemplazo de Parámetros ($\omega_0^2 = 1$, $Q = \frac{\sqrt{2}}{2}$)

Identificando los coeficientes de segundo orden estándar:

$$\omega_0^2 = \frac{1}{LC} = 1 \implies \omega_0 = 1\text{ rad/s}$$

$$\frac{\omega_0}{Q} = \frac{R}{L} = \frac{1}{\frac{\sqrt{2}}{2}} = \sqrt{2}$$

Sustituyendo estos valores en la función de transferencia:

$$H(s) = \frac{1}{s^2 + \sqrt{2}s + 1}$$


## 1.3. Polos y Ceros


El numerador es constante ($N(s) = 1$), por lo que **no posee ceros**.


Son las raíces del denominador $D(s) = 0$:

$$s^2 + \sqrt{2}s + 1 = 0$$

$$s_{1,2} = \frac{-\sqrt{2} \pm \sqrt{(\sqrt{2})^2 - 4(1)(1)}}{2} = \frac{-\sqrt{2} \pm \sqrt{-2}}{2}$$

$$s_{1,2} = -\frac{\sqrt{2}}{2} \pm j\frac{\sqrt{2}}{2}$$


## 1.4. Respuesta en Frecuencia: Módulo y Fase

Evaluando en el eje imaginario ($s = j\omega$):

$$H(j\omega) = \frac{1}{(j\omega)^2 + \sqrt{2}(j\omega) + 1} = \frac{1}{(1 - \omega^2) + j\sqrt{2}\omega}$$


$$|H(j\omega)| = \frac{|1|}{|(1 - \omega^2) + j\sqrt{2}\omega|} = \frac{1}{\sqrt{(1 - \omega^2)^2 + (\sqrt{2}\omega)^2}}$$

$$|H(j\omega)| = \frac{1}{\sqrt{1 - 2\omega^2 + \omega^4 + 2\omega^2}} = \frac{1}{\sqrt{1 + \omega^4}}$$


$$\phi(\omega) = \arg(1) - \arg\left[(1 - \omega^2) + j\sqrt{2}\omega\right] = -\arctan\left(\frac{\sqrt{2}\omega}{1 - \omega^2}\right)$$



 # 2. Circuito Pasa-altos



  $$Z_1(s) = R \parallel \frac{1}{sC} = \frac{R \cdot \frac{1}{sC}}{R + \frac{1}{sC}} = \frac{R}{sRC + 1}$$


  $$Z_2(s) = sL$$

Aplicando el divisor de tensión:

$$H(s) = \frac{V_2(s)}{V_1(s)} = \frac{Z_2(s)}{Z_1(s) + Z_2(s)} = \frac{sL}{\frac{R}{sRC + 1} + sL}$$

Multiplicando numerador y denominador por $(sRC + 1)$:

$$H(s) = \frac{sL(sRC + 1)}{R + sL(sRC + 1)} = \frac{s^2 RLC + sL}{s^2 RLC + sL + R}$$

Dividiendo numerador y denominador por $RLC$:

$$H(s) = \frac{s^2 + s\frac{1}{RC}}{s^2 + s\frac{1}{RC} + \frac{1}{LC}}$$



## 2.1 Reemplazo de Parámetros ($\omega_0^2 = 1$, $Q = \frac{\sqrt{2}}{2}$)


$$\omega_0^2 = \frac{1}{LC} = 1 \implies \omega_0 = 1\text{ rad/s}$$

$$\frac{\omega_0}{Q} = \frac{1}{RC} = \frac{1}{\frac{\sqrt{2}}{2}} = \sqrt{2}$$

Sustituyendo en la expresión de $H(s)$:

$$H(s) = \frac{s^2 + \sqrt{2}s}{s^2 + \sqrt{2}s + 1}$$



## 2.2 Polos y Ceros


Son las raíces del numerador $N(s) = 0$:

$$s(s + \sqrt{2}) = 0 \implies \begin{cases} z_1 = 0 \\ z_2 = -\sqrt{2} \end{cases}$$




Son las raíces del denominador $D(s) = 0$:

$$s^2 + \sqrt{2}s + 1 = 0$$

$$s_{1,2} = \frac{-\sqrt{2} \pm \sqrt{2 - 4}}{2} = -\frac{\sqrt{2}}{2} \pm j\frac{\sqrt{2}}{2}$$



## 2.3. Respuesta en Frecuencia: Módulo y Fase

Evaluando en el eje imaginario ($s = j\omega$):

$$H(j\omega) = \frac{(j\omega)^2 + \sqrt{2}(j\omega)}{(j\omega)^2 + \sqrt{2}(j\omega) + 1} = \frac{-\omega^2 + j\sqrt{2}\omega}{(1 - \omega^2) + j\sqrt{2}\omega}$$


$$|H(j\omega)| = \frac{|-\omega^2 + j\sqrt{2}\omega|}{|(1 - \omega^2) + j\sqrt{2}\omega|} = \frac{\sqrt{(-\omega^2)^2 + (\sqrt{2}\omega)^2}}{\sqrt{(1 - \omega^2)^2 + (\sqrt{2}\omega)^2}}$$

Desarrollando los términos algebraicos:

$$|H(j\omega)| = \frac{\sqrt{\omega^4 + 2\omega^2}}{\sqrt{1 - 2\omega^2 + \omega^4 + 2\omega^2}} = \frac{\omega\sqrt{\omega^2 + 2}}{\sqrt{\omega^4 + 1}}$$



Factorizando el numerador como $j\omega(j\omega + \sqrt{2})$:

$$\arg[N(j\omega)] = \arg(j\omega) + \arg(\sqrt{2} + j\omega) = \frac{\pi}{2} + \arctan\left(\frac{\omega}{\sqrt{2}}\right)$$

Restando la fase del denominador:

\begin{equation}
    \phi(\omega) = \frac{\pi}{2} + \arctan\left(\frac{\omega}{\sqrt{2}}\right) - \arctan\left(\frac{\sqrt{2}\omega}{1 - \omega^2}\right)
\end{equation}

# 3. Pasa-banda

  $$Z_1(s) = R$$


  $$Z_2(s) = sL \parallel \frac{1}{sC} = \frac{sL \cdot \frac{1}{sC}}{sL + \frac{1}{sC}} = \frac{sL}{s^2LC + 1}$$

Aplicando el divisor de tensión para obtener la relación entre $V_2(s)$ y $V_1(s)$

$$H(s) = \frac{V_2(s)}{V_1(s)} = \frac{Z_2(s)}{Z_1(s) + Z_2(s)} = \frac{\frac{sL}{s^2LC + 1}}{R + \frac{sL}{s^2LC + 1}}$$

Multiplicando numerador y denominador por $(s^2LC + 1)$:

$$H(s) = \frac{sL}{R(s^2LC + 1) + sL} = \frac{sL}{s^2 RLC + sL + R}$$

Dividiendo numerador y denominador por $RLC$ para llevarlo a su forma canónica:

$$H(s) = \frac{s \frac{1}{RC}}{s^2 + s\frac{1}{RC} + \frac{1}{LC}}$$


## 3.1 Reemplazo de Parámetros ($\omega_0^2 = 1$, $Q = \frac{\sqrt{2}}{2}$)



$$\omega_0^2 = \frac{1}{LC} = 1 \implies \omega_0 = 1\text{ rad/s}$$

$$\frac{\omega_0}{Q} = \frac{1}{RC} = \frac{1}{\frac{\sqrt{2}}{2}} = \sqrt{2}$$

Sustituyendo estos valores en la expresión de $H(s)$:

$$H(s) = \frac{\sqrt{2}s}{s^2 + \sqrt{2}s + 1}$$


## 3.2 Polos y Ceros

Son las raíces del numerador $N(s) = 0$:

$$\sqrt{2}s = 0 \implies z_1 = 0$$


Son las raíces del denominador $D(s) = 0$:

$$s^2 + \sqrt{2}s + 1 = 0$$

$$s_{1,2} = \frac{-\sqrt{2} \pm \sqrt{(\sqrt{2})^2 - 4(1)(1)}}{2} = \frac{-\sqrt{2} \pm \sqrt{-2}}{2}$$

$$s_{1,2} = -\frac{\sqrt{2}}{2} \pm j\frac{\sqrt{2}}{2}$$


## 3.3. Respuesta en Frecuencia: Módulo y Fase

Evaluando en el eje imaginario ($s = j\omega$):

$$H(j\omega) = \frac{j\sqrt{2}\omega}{(j\omega)^2 + \sqrt{2}(j\omega) + 1} = \frac{j\sqrt{2}\omega}{(1 - \omega^2) + j\sqrt{2}\omega}$$


$$|H(j\omega)| = \frac{|j\sqrt{2}\omega|}{|(1 - \omega^2) + j\sqrt{2}\omega|} = \frac{\sqrt{2}\omega}{\sqrt{(1 - \omega^2)^2 + (\sqrt{2}\omega)^2}}$$

$$|H(j\omega)| = \frac{\sqrt{2}\omega}{\sqrt{1 - 2\omega^2 + \omega^4 + 2\omega^2}} = \frac{\sqrt{2}\omega}{\sqrt{1 + \omega^4}}$$


$$\phi(\omega) = \arg[N(j\omega)] - \arg[D(j\omega)] = \frac{\pi}{2} - \arctan\left(\frac{\sqrt{2}\omega}{1 - \omega^2}\right)$$
